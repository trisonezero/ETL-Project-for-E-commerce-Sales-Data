```python
import pandas as pd
from sqlalchemy import create_engine

import json
```

```python
with open("config.json", "r") as f:
    c_data = json.load(f)

USER = c_data["USER"]
PASSWORD = c_data["PASSWORD"]
HOST = c_data["HOST"]
PORT = c_data["PORT"]
DATABASE = c_data["DATABASE"]

conn_string = f"postgresql://{USER}:{PASSWORD}@{HOST}:{PORT}/{DATABASE}"
engine = create_engine(conn_string)
```

```python
df = pd.read_csv(r"data_backup\orders.csv")
df.head(5)
```

```python
df.to_sql("orders_data", engine, if_exists="replace", index=False)
```

```python
df_test = pd.read_sql(f"SELECT * FROM orders_data;", engine)
df_test.head(5)
```
